```python
class human:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}さんは{self.age}歳なので成年です")
    else:
      print(f"{self.name}さんは{self.age}歳なのでまだ未成年です")


dic = {"太郎":104,"二郎":-2,"三郎":45,"四郎":77, "五郎":27}


for n,b in dic.items():
  human(n,b).check_adult()


```
