```python
class Human:
  def __init__(self,name,age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}さんは{self.age}歳なので成年です")
    else:
      print(f"{self.name}さんは{self.age}歳なのでまだ未成年です")


Humans = [Human("太郎",104),Human("二郎",-2),Human("三郎",45),Human("四郎",77), Human("五郎",27)]


for Human in Humans:
  Human.check_adult()
```
